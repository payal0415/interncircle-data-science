```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# E-commerce Sales Dataset
data = {
    "Date": [
        "2026-01-05","2026-01-12","2026-01-20","2026-02-03","2026-02-15",
        "2026-03-02","2026-03-18","2026-04-05","2026-04-22","2026-05-10",
        "2026-05-25","2026-06-08","2026-06-20","2026-07-05","2026-07-18",
        "2026-08-02","2026-08-15","2026-09-01","2026-09-12","2026-09-20"
    ],
    "Customer": [
        "C001","C002","C003","C004","C005",
        "C001","C006","C007","C002","C008",
        "C009","C003","C010","C004","C011",
        "C005","C012","C006","C013","C007"
    ],
    "Category": [
        "Electronics","Clothing","Home","Electronics","Beauty",
        "Clothing","Home","Electronics","Beauty","Clothing",
        "Electronics","Home","Beauty","Clothing","Electronics",
        "Home","Beauty","Electronics","Clothing","Home"
    ],
    "Quantity": [
        2,3,1,2,4,3,2,1,5,2,
        3,2,4,3,2,1,5,2,4,3
    ],
    "Sales": [
        24000,4500,3200,18000,6000,
        7500,5200,15000,5500,4200,
        28000,4800,7200,6800,22000,
        3900,8500,19500,7600,5100
    ]
}

df_sales = pd.DataFrame(data)

# Convert Date to datetime
df_sales["Date"] = pd.to_datetime(df_sales["Date"])

# Display dataset
print("First 5 Rows:")
display(df_sales.head())

print("\nDataset Information:")
df_sales.info()

print("\nSummary Statistics:")
display(df_sales.describe())

# Extract date features
df_sales["Month"] = df_sales["Date"].dt.month
df_sales["Month_Name"] = df_sales["Date"].dt.strftime("%B")

# Total sales
print("\nTotal Revenue:", df_sales["Sales"].sum())

# Sales by category
category_sales = df_sales.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print("\nSales by Category:")
display(category_sales)

# Top customers
customer_sales = df_sales.groupby("Customer")["Sales"].sum().sort_values(ascending=False)

print("\nTop Customers:")
display(customer_sales.head(5))
```

```python
# 1. Monthly Sales Trend

monthly_sales = df_sales.groupby("Month_Name")["Sales"].sum()

plt.figure(figsize=(10,5))
monthly_sales.plot(kind="bar")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()


# 2. Sales by Category

plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()


# 3. Top 5 Customers

plt.figure(figsize=(8,5))
customer_sales.head(5).plot(kind="bar")
plt.title("Top 5 Customers by Sales")
plt.xlabel("Customer")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()


# 4. Sales Trend Over Time

plt.figure(figsize=(10,5))
plt.plot(df_sales["Date"], df_sales["Sales"], marker="o")
plt.title("Sales Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
